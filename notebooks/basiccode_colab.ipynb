```
from pathlib import Path
import subprocess, sys
repo = Path('/content/unknown-exploration-basiccode')
if not repo.exists():
    subprocess.run(['git','clone','--branch','codex/basiccode-benchmark','https://github.com/sehyungpark/unknown-exploration-research.git',str(repo)],check=True)
subprocess.run([sys.executable,'-m','pip','install','-r','requirements-basiccode.txt'],cwd=repo,check=True)
```

```
RUN_ID = 'colab-check-001'
SMOKE = True
RESUME = False
args = [sys.executable,'-m','benchmarks.basiccode_runner','--run-id',RUN_ID,'--workers','2','--mode','isolated','--backup-root','/content/drive/MyDrive/basiccode-backups']
if SMOKE: args.append('--smoke')
if RESUME: args.append('--resume')
subprocess.run(args,cwd=repo,check=True)
```
